The following cell reads the text entries into Python. There are 96 entries in total for each of the following two questions:

1. Think of a specific situation in which someone challenged or supported a social stereotype or norm (refer to the first page of the survey for some examples) in your community. What happened, and how did this experience influence your personal views on social values? (min. 150 words)

2. Consider how Singapore, or your own home country, is adopting the use of AI. What are your thoughts on how this technology could impact the environment and society, and what concerns or hopes do you have about it? (min. 150 words)

```python
# Parse responses from text files (skip header and metadata)
def parse_responses(text):
    '''Extract responses separated by "--- Response N ---" markers'''
    lines = text.split('\n')
    responses = []
    current_response = []
    
    for line in lines:
        if line.startswith('--- Response'):
            if current_response:
                responses.append('\n'.join(current_response).strip())
            current_response = []
        elif line.strip() and not line.startswith('==='):
            current_response.append(line)
    
    if current_response:
        responses.append('\n'.join(current_response).strip())
    
    return [r for r in responses if len(r) > 50]  # Filter out very short/empty responses

social_norm_responses = parse_responses(cw_text)
ai_responses = parse_responses(cx_text)

print(f"Loaded {len(social_norm_responses)} social norm responses")
print(f"Loaded {len(ai_responses)} AI responses")
print(f"\nFirst social norm response (sample):\n{social_norm_responses[0][:200]}...")
```
