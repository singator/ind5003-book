# Worksheet 1: Familiarisation with Pandas and EDA

Exploratory Data Analysis (EDA) is a pivotal part of any data analysis. Although we do not have a dedicated chapter on it in our course,
it *must* be present in your report. EDA provides clues on how to approach your problem, in its specific context. Data analysts who do not 
perform a thorough EDA are reduced to trying model after model, and will not be able to provide custom recommendations.

EDA consists of studying graphical and numerical summaries of the data, perhaps conditioned on important/relevant sub-groups. In this worksheet, 
we introduce a tool for generating such summaries automatically. To begin, ensure `sweetviz` is installed in your virtual environment, and run it on the happiness report.

> Let's start with a basic aim of understanding how happiness score is related to the other indicators.

1. Run `sweetviz` on data and study the output. [Documentation link](https://github.com/fbdesignpro/sweetviz)

```python
import numpy as np
import pandas as pd
import sweetviz as sv
import matplotlib.pyplot as plt
import seaborn as sns
from itables import show

sns.set_theme(style="whitegrid")
```

```python
happ = pd.read_csv('data/happiness_report.csv', header=0, na_values='NA')
happ.head()
```

The following generates a report that quickly provides an overview of the relationship between the variables.

```python
my_report = sv.analyze(happ)
my_report.show_notebook() # Default arguments will generate to "SWEETVIZ_REPORT.html"
```
