### Indexing and Selecting Data

```python
print(happ.loc[2:5])
```

```python
#| label: tbl-happ2
#| tbl-cap: Extracting a subset of columns using the slice operator with .loc

happ.loc[2:10:4, "GDP":"Generosity":2]
```

```python
#| label: tbl-happ3
#| tbl-cap: Extracting a subset of columns using the slice operator with .iloc

happ.iloc[2:11:4, 3:8:2] 
```

### Filtering Data
### Example: Happiest countries

```python
happiest = happ[happ['Happiness.Score'] > 6.95]

happiest.Country.unique()
```

```python
# Top 3 happiest countries in 2015
print(happ[(happ.Year == 2015) & (happ['Happiness.Rank'] <= 3)])
```
